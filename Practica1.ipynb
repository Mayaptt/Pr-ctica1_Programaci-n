```python
#Variables
paquete1 = {
    "ID":"P0001",
    "peso":4
    }
paquete2 = {
    "ID":"P0002",
    "peso":6
    }

fifo = [paquete1, paquete2]
lifo = [paquete1, paquete2]
```

```python
menu = True
while menu:

    #Menú
    print("\nBienvenido al Sistema Avanzado FastBox")
    print("1. Registrar nuevo paquete")
    print("2. Despachar paquete")
    print("3. Ver estado detallado de las rutas")
    print("4. Salir")
    print("===========================================")
    opcion = input("Elige una opción (1-4): ")
    
    if (opcion.isnumeric()):
    
        #Elección de la opción
        match int(opcion):
            
            case 1:#Registrar nuevo paquete
                nuevo = input()
                
            case 2:#Despachar paquete
                repetir = True
                while repetir:    
                    ruta = input ("¿De qué ruta quieres despachar? (1-Estándar / 2-Urgente): ")
                    if (ruta.isnumeric()):
                        if int(ruta) == 1:#Ruta estándar
                            if fifo:
                                print(f"[ÉXITO] Estándar - FIFO Entregado paquete {fifo [0] ["ID"]} de {fifo [0] ["peso"]} kg.")
                                repetir = False
                            else:
                                otra_ruta = input("\nNo hay paquetes en la ruta estándar. ¿Quieres revisar otra ruta? (Sí/no): ")
                                if otra_ruta == "no" or otra_ruta == "n" or otra_ruta == "NO" or otra_ruta == "No":
                                    repetir = False
                                
                        elif int(ruta) == 2:#Ruta urgente
                            if lifo:
                                lifo2 = lifo [::-1]
                                print(f"[ÉXITO] Urgente - LIFO Entregado paquete {lifo2 [0] ["ID"]} de {lifo2 [0] ["peso"]} kg.")
                                repetir = False
                            else:
                                otra_ruta = input("\nNo hay paquetes en la ruta urgente. ¿Quieres revisar otra ruta? (Sí/no): ")
                                if otra_ruta == "no" or otra_ruta == "n" or otra_ruta == "NO" or otra_ruta == "No":
                                    repetir = False
                        else:
                            print("[ERROR] Ruta no válida. Por favor, introduce 1 o 2.")
                    else:
                        print("[ERROR] Por favor, introduce 1 o 2.")
                
            case 3:#Ver estado detallado de las rutas
                print("\n--- ESTADO ACTUAL DE LAS RUTAS ---")
                print("\nRuta estándar:")
                for elemento in fifo:
                    print(f"ID: {elemento ["ID"]} | Peso: {elemento ["peso"]} kg")
                print("\nRuta urgente:")
                indice = len(lifo)
                while indice != 0:
                    print(f"ID: {lifo [indice-1] ["ID"]} | Peso: {lifo [indice-1] ["peso"]} kg")
                    indice -= 1
                    
            case 4:#Salir
                print("¡Hasta luego!")
                menu = False
                
            case _:#Cualquier otra opción
                print("[ERROR] Opción no válida. Por favor, introduce un número del 1 al 4.")
    else:
        print("[ERROR] Opción no válida. Por favor, introduce un número del 1 al 4.")
```
